```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
X, _= make_blobs(n_samples = 1000, centers = 3, n_features=2,random_state = 42)
mbk_means = MiniBatchKMeans(n_clusters=3, random_state=42)
mbk_means.fit(X)

plt.scatter(X[:,0], X[:,1], c=mbk_means.labels_ ,cmap='viridis')
plt.scatter(mbk_means.cluster_centers_[:,0], mbk_means.cluster_centers_[:,1], s=300, c='r', marker='x')
plt.title('Mini-Batch K-Means Clustering')
plt.show()
```
